```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# European option files
call_path = 'european_bs_call_solution.txt'
put_path  = 'european_bs_put_solution.txt'

# Load (skip header)
S_call, V_call = np.loadtxt(call_path, skiprows=1, unpack=True)
S_put,  V_put  = np.loadtxt(put_path,  skiprows=1, unpack=True)

# Plot
plt.figure(figsize=(8,5))
plt.plot(S_call, V_call, label='European Call', color='tab:blue', linewidth=2)
plt.plot(S_put,  V_put,  label='European Put',  color='tab:orange', linewidth=2)
plt.xlabel('Spot Price')
plt.ylabel('Option Value')
plt.title('European Black–Scholes Prices Computed via Crank–Nicolson Method')
plt.grid(True, alpha=0.4)
plt.legend( loc='upper center')
plt.tight_layout()
plt.savefig('bs_prices.png', dpi=200)
plt.show()
```
