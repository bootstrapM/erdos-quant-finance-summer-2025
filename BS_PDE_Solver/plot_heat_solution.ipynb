```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load data (skip header)
file_path = 'heat_eq_sol.txt'
x, u = np.loadtxt(file_path, skiprows=1, unpack=True)

# Plot
plt.figure(figsize=(7,4))
plt.plot(x, u, color='tab:blue', linewidth=2)
plt.xlabel('x')
plt.ylabel('u(x,t=1)')
plt.title('Solution of 1d heat equation at t=1 obtained via Crank–Nicolson method')
plt.grid(True, alpha=0.4)
plt.tight_layout()

# Save a copy to file and display
plt.savefig('heat_sol.png', dpi=200)
plt.show()
```
